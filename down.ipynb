```python
# Change only this URL, then run all cells.
URL = "https://www.youtube.com/watch?v=KVHtuwVhULA"

OUTPUT_DIR = "youtube_subs"
MODEL = "medium"
LANGUAGE = ""
TRANSLATE_TO = "ko"
CONCEPT_STYLE = "ko_en"
DEVICE = "cuda"
COMPUTE_TYPE = "float16"
BEAM_SIZE = 5
PROMPT_HINT = ""
KEEP_AUDIO = False
```

```python
from youtube_subtitle_cli import run_job

result = run_job(
    url=URL,
    output_dir=OUTPUT_DIR,
    model=MODEL,
    language=LANGUAGE,
    device=DEVICE,
    compute_type=COMPUTE_TYPE,
    beam_size=BEAM_SIZE,
    prompt_hint=PROMPT_HINT,
    translate_to=TRANSLATE_TO,
    concept_style=CONCEPT_STYLE,
    keep_audio=KEEP_AUDIO,
)

print(result["info"].get("title"))
result["outputs"]
```
